## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which pages a content reviewer should look at first, given limited time
to review thousands of pages.

Who acts, and how: a content reviewer opens the top-ranked pages from the queue and
decides to refresh, expand, protect, prune, or monitor each one.

Cost of a wrong call: a false positive wastes a reviewer's limited hours on a page
that didn't need attention. A false negative is more expensive — a genuinely
declining, high-demand page keeps losing visibility unreviewed until someone
notices some other way.

Why a plain rule isn't enough: a hand-written rule (e.g. "180+ days stale AND 500+
impressions") only flags 17 pages out of 30,000 here — far too narrow, and it can't
weigh several weak, tangled signals (age, position, CTR, freshness, engagement)
together the way a learned model can.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd, json

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
res = json.load(open("outputs/model_results.json"))

declining_rate = (df["trend_direction"] == "down").mean()
stale_visible_count = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
declining_with_demand = ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum()

print(f"Rows: {len(df)}")
print(f"Declining rate (trend_direction == 'down'): {declining_rate:.3f}")
print(f"Pages that are both stale (180+ days) and visible (500+ impressions): {stale_visible_count}")
print(f"Declining pages with real demand (impressions_90d >= 100): {declining_with_demand}")
print(f"Baseline rule Precision@50: {res['baseline']['baseline_precision_at_50']}")
print(f"Random forest Precision@50: {res['models']['random_forest']['precision_at_50']}")
```

```python
print("""
These numbers show a specific, sizeable problem, not a vague one: 54% of pages show
some decline, and 13,152 combine decline with real demand — but a strict hand-written
rule only surfaces 17 pages. That gap between rule and reality is exactly what this
lane exists to close, and the baseline-vs-model comparison (0.24 -> 0.74 Precision@50)
is direct evidence a learned ranking does better here.
""")
```
